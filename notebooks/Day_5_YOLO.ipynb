#  What is YOLO?

**YOLO (You Only Look Once)** is one of the **fastest** and most popular **real-time object detection algorithms** used in Computer Vision.

Unlike older object detection methods that scan an image multiple times, **YOLO processes the entire image in a single pass (one look)**, making it extremely fast while maintaining high accuracy.

---

## 🔍 YOLO Predicts Three Things

For every image, YOLO predicts:

- 📦 **What objects are present?**
  - Identifies the class of the object (e.g., Person, Car, Dog, Bottle).
  - 
    

- 📍 **Where are they located?**
  - Predicts the object's location using a **Bounding Box**.
  - 
    

- 🎯 **How confident is the prediction?**
  - Provides a **Confidence Score** indicating how certain the model is about its prediction.

---

## 💡 Why is YOLO So Fast?

- ✅ Looks at the **entire image only once**
- ✅ Performs **classification and localization simultaneously**
- ✅ Can process **real-time video streams**
- ✅ Suitable for applications like surveillance, autonomous driving, robotics, and face detection

---

## 📊 Traditional Methods vs YOLO

| Traditional Object Detection | YOLO |
|------------------------------|------|
| Multiple passes over the image | Single pass over the image |
| Slower | Extremely Fast |
| Complex pipeline | End-to-end detection |
| Suitable for offline tasks | Ideal for Real-Time Applications |

---

## 🖼️ YOLO Workflow

```text
Input Image
      │
      ▼
Image is divided into Grid Cells
      │
      ▼
Neural Network processes the entire image
      │
      ▼
Predicts:
 • Object Class
 • Bounding Box
 • Confidence Score
      │
      ▼
Final Detected Objects
```

## Loading Libraries

````python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
import numpy as np
````

#

## Object Detection using YOLO

````python
model = YOLO("yolov8n.pt")
````

###  Object Detection on Image

````python
image = cv2.imread("marvel.jpg")
results = model(image)
````

````python
annotated_image = results[0].plot()
plt.figure(figsize=(10,8))
plt.imshow(cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
````

###  Object Detection on Webcam


````python
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    
    if not ret:
        break

    
    resized_frame = cv2.resize(frame, (1080, 600))    

    results = model(resized_frame)
    annotated_frame = results[0].plot()
    
    cv2.imshow("YOLO Live Detection", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()

````
